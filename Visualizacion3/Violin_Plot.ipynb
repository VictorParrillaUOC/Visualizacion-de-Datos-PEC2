```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

df = pd.read_csv("StudentsPerformance.csv")
df.head()
```

```python
df_long = df.melt(
    id_vars="gender",
    value_vars=["math score", "reading score", "writing score"],
    var_name="subject",
    value_name="score"
)

sns.violinplot(
    data=df_long,
    x="subject",
    y="score",
    hue="gender",
    inner="quartile",
    split=True
)

plt.title("Notas de estudiantes por asignatura y sexo")
plt.xlabel("Asignatura")
plt.ylabel("Score")
plt.tight_layout()
plt.savefig('Visualizacion3/Violin_Plot.png', dpi=300, bbox_inches="tight")
plt.show()
```
